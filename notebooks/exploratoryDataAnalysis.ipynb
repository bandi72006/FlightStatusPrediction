## Preprocessing

Get rid of any post-flight information (would cause data leaks) + redundant info

```python
postFlightCols = [
    "DepTime", "DepDelay", "DepDelayMinutes",
    "ArrTime",                                    
    "ArrDelayMinutes",                            
    "ActualElapsedTime",                          
    "WheelsOff", "WheelsOn",                      
    "ArrivalDelayGroups",                         
    "DivAirportLandings"
]

redundantCols = ["DOT_ID_Marketing_Airline", "IATA_Code_Marketing_Airline",
    "DOT_ID_Operating_Airline", "IATA_Code_Operating_Airline",
    "OriginAirportSeqID", "OriginCityName", "OriginStateName",
    "DestAirportSeqID", "DestCityName", "DestStateName",
    "Year"]

dropCols = postFlightCols + redundantCols

df.drop(columns=dropCols, inplace=True) # inplace=True -> applies changes to actual df


print(df.shape)
df.head(1)
```

### Feature engineering

```python
from sklearn.preprocessing import LabelEncoder

# Handle missing values

df.fillna({"CRSElapsedTime":df["CRSElapsedTime"].median()}, inplace=True)
df.fillna({"Tail_Number": "None"}, inplace=True)
df.fillna({"Operated_or_Branded_Code_Share_Partners": "None"}, inplace=True)
df.fillna({"ArrTimeBlk": df["ArrTimeBlk"].mode()[0]}, inplace=True)
df.fillna({"CRSDepTime": 0}, inplace=True)
df.fillna({"CRSArrTime": 0}, inplace=True) 

# Columns to engineer
viewCols = ["CRSDepTime", "CRSArrTime", "DayOfWeek", "Month"]
print(df[viewCols].head(5))

df["DepHour"] = (df["CRSDepTime"].astype(int) // 100).clip(0,23)
df["ArrHour"] = (df["CRSArrTime"].astype(int) // 100).clip(0,23)
df["IsWeekend"] = (df["DayOfWeek"] >= 6).astype(int)

seasonMap = {12: 0,1:"Winter",2:"Winter",3:"Spring",4:"Spring",5:"Spring",
              6:"Summer",7:"Summer",8:"Summer",9:"Fall",10:"Fall",11:"Fall"}
df["Season"] = df["Month"].map(seasonMap)
if "Season_Summer" not in df.columns and "Season_Spring" not in df.columns and "Season_Winter" not in df.columns: # No fall season in dataset
    df = pd.get_dummies(df, columns=["Season"]) # One-hot encodes feature

le = LabelEncoder()
df["Airline"] = le.fit_transform(df["Airline"]) # Encodes airline to a digit between 0 and numberOfAirlines-1

# Y-values

conditions = [df["Cancelled"] == True, df["Diverted"] == True, df["DepDel15"] == 1.0]
choices = ["Cancelled", "Diverted", "Delayed"]
df["DelayStatus"] = np.select(conditions, choices, default="OnTime")

# Convert boolean to ints

boolCols = df.select_dtypes(include="bool").columns
df[boolCols] = df[boolCols].astype(int)

print(df.shape)
print(df.head(1))

```

### Correlation

```python
# Only predicting with these features

featureCols = [
    'DepHour', 'ArrHour', 'DayOfWeek', 'DayofMonth', 'Month', 'Quarter',
    'IsWeekend', 'Season_Summer', 'Season_Winter', 'Season_Spring',
    'Distance', 'CRSElapsedTime', 'DistanceGroup',
    'Airline', 
    'OriginAirportID', 'DestAirportID',
    'OriginCityMarketID', 'DestCityMarketID',
    'OriginWac', 'DestWac',
    'OriginStateFips', 'DestStateFips'
]

featureDf = df[featureCols + ["DepDel15"]].copy()

corrMatrix = featureDf.corr(numeric_only=True)

delayCorr = corrMatrix[["DepDel15"]].sort_values(by="DepDel15", ascending=False)

plt.figure(figsize=(4,12))
sns.heatmap(delayCorr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation of features with flight delay')
plt.show()
```

- Not much correlation between whether a flight is delayed and features
- Shows relationship between features and delay is not a simple linear one

- Departure hour and arrival hour affects flights delay the most - 2 highest correlations

## Machine learning

### Prepare data

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import lightgbm as lgb
from sklearn.metrics import classification_report
import pandas as pd
import time
```

```python
# Only keeps flights that are not cancelled or diverted

dfDelay = df[(df["Cancelled"] == False) & (df["Diverted"] == False)].copy()
dfDelay = dfDelay.dropna(subset=["DepDel15"]) 


counts = dfDelay["DelayStatus"].value_counts()
print(counts)

print(f"Data split: \nOnTime: {counts["OnTime"]/dfDelay.shape[0]*100:.2f}%, Delayed: {counts["Delayed"]/dfDelay.shape[0]*100:.2f}% \n")


X = dfDelay[featureCols].copy()
y = dfDelay["DepDel15"].copy()

print("Total data shape: ")
print(X.shape, y.shape, "\n")

RANDOMSTATE = 42
XTrain, XTest, yTrain, yTest = train_test_split(X, y, train_size=0.1, test_size=0.2, random_state=RANDOMSTATE, stratify=y) # Stratify = ensures test-train split has same ratio of classes after split

print("Train data shape: ")
print(XTrain.shape, yTrain.shape)
print("Test data shape: ")
print(XTest.shape, yTest.shape)
```

Data imbalance: uneven number of samples for each class, may cause classification issues

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from pickle import dump, load
import os

if not os.path.isfile("../models/rf.pkl"):
    rf = RandomForestClassifier(
        n_estimators = 100,
        class_weight = "balanced", # Handles data sample imbalanace
        random_state = RANDOMSTATE,
        n_jobs = 2, # Tells CPU to use ALL available CPU cores
        verbose = 2
    )

    
    rf.fit(XTrain, yTrain)

    with open("../models/rf.pkl", "wb") as f:
        dump(rf, f, protocol=5)
        print("Model trained")

else:
    with open("../models/rf.pkl", "rb") as f:
        rf = load(f)
        print("Model loaded")

```

```python
THRESHOLD = 0.3  # lower than 0.5 to boost recall (see next to choose the right value)

proba = rf.predict_proba(XTest)[:, 1] # Get probability for only one class not both via array slicing
yPred = (proba >= THRESHOLD).astype(int)

print(classification_report(yTest, yPred, digits=3))
```

### Finding the best threshold to have the best accuracy and precision tradeoffs

```python
from sklearn.metrics import precision_score, recall_score, f1_score

def thresholdFinder(proba):
    print("Thresh Prec_1 Rec_1 F1_1")
    for threshold in [0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
        thresholdPred = (proba >= threshold).astype(int)
        precision = precision_score(yTest, thresholdPred, pos_label=1)
        recall = recall_score(yTest, thresholdPred, pos_label=1)
        f1 = f1_score(yTest, thresholdPred, pos_label=1)
        print(f"{threshold}: {precision:.3f}, {recall:.3f}, {f1:.3f}")
    
```

```python
thresholdFinder(proba)
```

### XGBoost

```python
from xgboost import XGBClassifier
from pickle import dump, load
import os

scalePosWeight = (yTrain == 0).sum() / (yTrain == 1).sum() # Used to adjust for dataset imbalance

if not os.path.isfile("../models/xgb.pkl"):

    xgb = XGBClassifier(
        n_estimators = 100,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=RANDOMSTATE,
        n_jobs=2,
        scale_pos_weight = scalePosWeight,
        eval_metric="logloss"
    )

    xgb.fit(XTrain, yTrain)


    with open("../models/xgb.pkl", "wb") as f:
        dump(rf, f, protocol=5)
        print("Model trained")

else:
    with open("../models/xgb.pkl", "rb") as f:
        lgbm = load(f)
        print("Model loaded")
```

```python
proba = xgb.predict_proba(XTest)[:, 1]
yPred = (proba >= THRESHOLD).astype(int)
print(classification_report(yTest, yPred, digits=3))
```

## Hyperparameter Tuning

```python
import optuna

def optunaObjective(trial): # Objective function for Optuna
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 50, 100, 300),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.2),
        "max_depth": trial.suggest_int("max_depth", 3, 10),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 10),
        "gamma": trial.suggest_float("gamma", 0, 5),
        "reg_alpha": trial.suggest_float("reg_alpha", 0, 5),
        "reg_lambda": trial.suggest_float("reg_lambda", 0, 5),
        "random_state": RANDOMSTATE,
        "n_jobs": 2,
        "scale_pos_weight": (yTrain == 0).sum() / (yTrain == 1).sum(),
        "eval_metric": "logloss"
    }

    model = XGBClassifier(**params)
    model.fit(XTrain, yTrain)

    proba = xgb.predict_proba(XTest)[:, 1]
    yPred = (proba >= THRESHOLD).astype(int)

    return recall_score(yTest, yPred, pos_label=1) 

study = optuna.create_study(direction="maximize")
study.optimize(optunaObjective, n_trials=30)


print("Best parameters:", study.best_params)
print("Best recall:", study.best_value)
```
